Test SPARQL INSERT auf das Dataset ausführen

```python
from SPARQLWrapper import SPARQLWrapper, JSON
sparql = SPARQLWrapper(f"{ontodocker_url}/api/v1/jena/{test_dataset_name}/sparql")
sparql.addCustomHttpHeader("Authorization", f"Bearer {ontodocker_jwt}")
sparql.setReturnFormat(JSON)
# INSERT
insert_query = """
        PREFIX dc: <http://purl.org/dc/elements/1.1/>
        INSERT DATA
        { 
            <http://example/book1> dc:title "A new book" .
        }
        """
sparql.setMethod("POST")
sparql.setQuery(insert_query)
results = sparql.query().convert()
print(results)
```

Abfragen des Datasets mit einem SPARQL Query

```python
select_query = "SELECT ?s ?p ?o WHERE {?s ?p ?o}"
sparql.setMethod("GET")
sparql.setQuery(select_query)
sparql.setReturnFormat(JSON)
results = sparql.query().convert()
for result in results["results"]["bindings"]:
    print(result["s"], result["p"], result["o"])
```
